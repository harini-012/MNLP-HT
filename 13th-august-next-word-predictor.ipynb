```python
import re
import math
import numpy as np
import nltk
from collections import Counter,defaultdict
from datasets import load_dataset
nltk.download("punkt")
print("Loading WikiText-2 dataset...")
dataset=load_dataset("Salesforce/wikitext", "wikitext-2-raw-v1")
text="\n".join(dataset["train"]["text"])
print("Dataset Loaded Successfully!")
def preprocess_text(text):
    text=text.lower()
    text = re.sub(r"[^a-z0-9\s]", " ", text)
    text=re.sub(r"\s+"," ",text)
    return text.strip()
clean_text=preprocess_text(text)
tokens=nltk.word_tokenize(clean_text)
print("Total Tokens:",len(tokens))
tokens=tokens[:100000]
print("Tokens used for training:",len(tokens))
unigram_counts=Counter(tokens)
bigram_counts=Counter()
trigram_counts=Counter()
for i in range(len(tokens)-1):
    bigram=(tokens[i],tokens[i+1])
    bigram_counts[bigram]+=1
for i in range(len(tokens)-2):
    trigram=(tokens[i],tokens[i+1],tokens[i+2])
    trigram_counts[trigram]+=1
vocabulary=set(tokens)
vocab_size=len(vocabulary)
print("Vocabulary size:",vocab_size)
def predict_next_word(sentence,top_n=5,k=1):
    sentence=preprocess_text(sentence)
    words=sentence.split()
    if len(words)==0:
        return []
    candidates=[]
    if len(words)>=2:
        previous_word_1=words[-2]
        previous_word_2=words[-1]
        denominator=bigram_counts[(previous_word_1,previous_word_2)]
        for word in vocabulary:
            numerator=trigram_counts[(previous_word_1,previous_word_2,word)]
            probability=(numerator+k)/(denominator+k*vocab_size)
            score=math.log(probability)
            candidates.append((word,score))
    else:
        previous_word=words[-1]
        denominator=unigram_counts[previous_word]
        for word in vocabulary:
            numerator=bigram_counts[(previous_word,word)]
            probability=(numerator+k)/(denominator+k*vocab_size)
            score=math.log(probability)
            candidates.append((word,score))
    candidates.sort(key=lambda x:x[1],reverse=True)
    return candidates[:top_n]
print("\n======================================")
print("SMART NEXT WORD PREDICTOR")
print("Type 'exit' to stop.\n")
while True:
    s=input("Enter the sentence:")
    if s.lower()=="exit":
        print("Program stopped.")
        break
    predictions=predict_next_word(s,top_n=5,k=1)
    if not predictions:
        print("Please enter some words")
        continue
    print("\nNext word suggestions:")
    for i,(word,score) in enumerate(predictions,start=1):
        print(f"{i}.{word}")
    print()
```
